# Practical 21: Decision Tree Regression

**Aim:** To predict employee salary based on experience level using Decision Tree Regression.

**Dataset:** Data Science Job Salaries (Kaggle)

**Kaggle link:** https://www.kaggle.com/datasets/ruchi798/data-science-job-salaries

**Dataset setup:**
1. Download and extract the Kaggle dataset.
2. Keep `ds_salaries.csv` in the same folder as this notebook.
3. Run each cell using **Shift + Enter**.

**Columns used:** `experience_level` and `salary_in_usd`. Experience levels are converted into numbers: EN=1, MI=2, SE=3, EX=4.

This follows your uploaded Practical 3 reference: load data, display data, separate X and y, scatter plot, initialize model, `fit()`, `predict()`, `score()` and graph.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
```

## 2. Load the dataset

```python
df = pd.read_csv("ds_salaries.csv")
df.head()
```

## 3. Display the dataset

```python
df.info()
```

```python
df.describe()
```

## 4. Separate X and y

```python
df["experience_level"] = df["experience_level"].map({"EN": 1, "MI": 2, "SE": 3, "EX": 4})

X = df[["experience_level"]]
y = df["salary_in_usd"]

print(X.head())
print(y.head())
```

## 5. Display the data using scatter plot

```python
plt.title("Salary vs Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Salary in USD")
plt.scatter(X, y, color="red")
plt.show()
```

## 6. Initialize the algorithm

```python
model = DecisionTreeRegressor(random_state=0)
```

## 7. Train the model using fit()

```python
model.fit(X, y)
```

## 8. Predict salary using predict()

```python
# Predict salary for Senior level (SE = 3)
ip = [[3]]
model.predict(ip)
```

## 9. Display accuracy

```python
print("Accuracy:", model.score(X, y) * 100, "%")
```

## 10. Display the Decision Tree Regression graph

```python
plt.title("Decision Tree Regression")
plt.xlabel("Experience Level")
plt.ylabel("Salary in USD")
plt.scatter(X, y, color="red")
plt.plot([1, 2, 3, 4], model.predict([[1], [2], [3], [4]]), color="blue")
plt.show()
```

## Conclusion

Thus, Decision Tree Regression was implemented to predict employee salary based on experience level. The model was trained using `fit()` and salary was predicted using `predict()`.
